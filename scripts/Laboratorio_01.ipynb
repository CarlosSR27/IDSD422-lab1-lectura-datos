## **2. Leer un JSON con python**

**TODO 1: importa json y pandas**

```python
import json
import pandas as pd
```

**TODO 2: carga data/ventas.json con json.load**

```python
with open('data/ventas.json', encoding="utf-8") as f:
  datos = json.load(f)

  # Imprime las primeras filas
  ventas = pd.json_normalize(datos)
  print(ventas.head())

  # Las dimenciones de la tabla
  print ('\nDimenciones',ventas.shape)

  # Imprime los tipos de columna
  print('\n Tipos por columna' )
  print(ventas.dtypes)
```
